```python
import os
from openai import OpenAI
import datetime

client = OpenAI(api_key=os.getenv("voc-1222044512175350470816369f4bca05d9987.79993304"))

# ----------- Prompt Strategies -----------

def basic_prompt(user_query):
    return f"""
You are a banking assistant. Answer the user query.

User: {user_query}
"""


def structured_prompt(user_query):
    return f"""
You are a Retail Banking Loan Advisory Assistant.

Follow rules:
- Do NOT approve or reject loans
- Provide eligibility guidance only
- Highlight risks clearly
- Ask for missing information
- Be concise and structured

Format:
Eligibility: <Eligible/Borderline/Not Eligible/Need More Info>
Reason:
Risks:
Next Steps:

User Query: {user_query}
"""


def chain_of_thought_prompt(user_query):
    return f"""
You are an expert loan officer.

Think step-by-step before answering:
1. Identify user intent
2. Extract financial details
3. Assess eligibility
4. Identify risks
5. Provide recommendation

Then provide final structured answer.

User Query: {user_query}
"""


# ----------- LLM Call -----------

def call_llm(prompt):
    response = client.chat.completions.create(
        model="gpt-4.1-mini",  # cost-effective
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3
    )
    return response.choices[0].message.content


# ----------- Agent -----------

def loan_agent_llm(user_input, strategy="structured"):
    
    if "approve loan" in user_input.lower():
        return "I cannot approve loans. Please contact a bank officer."

    if strategy == "basic":
        prompt = basic_prompt(user_input)
    elif strategy == "cot":
        prompt = chain_of_thought_prompt(user_input)
    else:
        prompt = structured_prompt(user_input)

    response = call_llm(prompt)
    return response


# ----------- Logging -----------

def log_interaction(user_input, response, strategy):
    with open("phase3_logs.txt", "a") as f:
        f.write(f"\n[{datetime.datetime.now()}]")
        f.write(f"\nStrategy: {strategy}")
        f.write(f"\nUser: {user_input}")
        f.write(f"\nAgent: {response}\n")


# ----------- Run -----------

def run_agent():
    print("=== LLM Loan Agent ===")
    print("Choose strategy: basic / structured / cot")

    strategy = input("Strategy: ")

    while True:
        user_input = input("\nYou: ")

        if user_input.lower() == "exit":
            break

        response = loan_agent_llm(user_input, strategy)
        print("\nAgent:", response)

        log_interaction(user_input, response, strategy)


if __name__ == "__main__":
    run_agent()
```
