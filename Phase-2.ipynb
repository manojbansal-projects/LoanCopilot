```python
import datetime

# ----------- Helper Functions -----------

def calculate_emi(loan_amount, interest_rate=10, tenure_years=5):
    r = interest_rate / (12 * 100)
    n = tenure_years * 12
    emi = loan_amount * r * (1 + r)**n / ((1 + r)**n - 1)
    return round(emi, 2)


def check_eligibility(income, credit_score):
    if income >= 30000 and credit_score >= 700:
        return "Eligible"
    elif income >= 20000 and credit_score >= 650:
        return "Borderline"
    else:
        return "Not Eligible"


def detect_risk(credit_score, income):
    risks = []
    if credit_score < 650:
        risks.append("Low credit score")
    if income < 25000:
        risks.append("Low income")
    return risks


# ----------- Agent Core Logic -----------

def loan_agent(user_input):
    user_input = user_input.lower()

    # Rule 1: Eligibility Check
    if "eligible" in user_input:
        try:
            income = int(input("Enter monthly income: "))
            credit_score = int(input("Enter credit score: "))
            
            status = check_eligibility(income, credit_score)
            risks = detect_risk(credit_score, income)

            response = f"Eligibility Status: {status}\n"
            if risks:
                response += f"Risks: {', '.join(risks)}"
            else:
                response += "No major risks detected"

            return response
        
        except:
            return "Invalid input. Please enter numeric values."

    # Rule 2: EMI Calculation
    elif "emi" in user_input:
        try:
            loan_amount = int(input("Enter loan amount: "))
            tenure = int(input("Enter tenure (years): "))
            
            emi = calculate_emi(loan_amount, tenure_years=tenure)
            return f"Estimated EMI: ₹{emi}"
        
        except:
            return "Invalid input for EMI calculation."

    # Rule 3: Document Checklist
    elif "documents" in user_input:
        return ("Required Documents:\n"
                "- ID Proof\n"
                "- Address Proof\n"
                "- Income Proof\n"
                "- Bank Statements")

    # Rule 4: Restricted Action (Safety)
    elif "approve loan" in user_input:
        return "I cannot approve loans. Please contact a bank officer."

    # Default Response
    else:
        return "Sorry, I can help with eligibility, EMI, or documents."


# ----------- Logging -----------

def log_interaction(user_input, response):
    with open("agent_logs.txt", "a") as f:
        f.write(f"\n[{datetime.datetime.now()}]")
        f.write(f"\nUser: {user_input}")
        f.write(f"\nAgent: {response}\n")


# ----------- Run Agent -----------

def run_agent():
    print("=== Loan Origination Assistant (Baseline) ===")
    print("Type 'exit' to quit\n")

    while True:
        user_input = input("You: ")
        
        if user_input.lower() == "exit":
            print("Goodbye!")
            break

        response = loan_agent(user_input)
        print("Agent:", response)

        log_interaction(user_input, response)


if __name__ == "__main__":
    run_agent()
```
